```python

import pandas as pd

data = []

for row in ws.iter_rows(min_row=2, max_row=ws.max_row, min_col=1, max_col=13):
    excel_row = row[0].row
    values= [cell.value for cell in row]
    data.append([excel_row] + values)

df = pd.DataFrame(data, columns=["excel row", "A", "B", "C", "D", "E", "F","G","H","I","J","K","L","M"])
df.dtypes
df.head()
```

```python
empty_cells = df[['C','J']].isnull().any(axis=1)
display(empty_cells)
empty_cells.sum()
display(df[empty_cells])
```

```python
clean_empty_cells = df[~empty_cells]
clean_empty_cells
```
